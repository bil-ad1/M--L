```python
import pandas as pd
```

```python
epx= pd.read_csv('/content/empz_01.csv')
```

```python
epz = epx.dropna()
```

```python
epz.info()
```

```python
from sklearn.preprocessing import LabelEncoder
l_e = LabelEncoder()
```

```python

dep_enc = l_e.fit_transform(epz.Department)
biz_tvl = l_e.fit_transform(epz.BusinessTravel)
atr = l_e.fit_transform(epz.Attrition)
```

```python
X = list(zip(dep_enc,biz_tvl))
```

```python
y = atr
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
mod = KNeighborsClassifier(n_neighbors=3)
```

```python
mod.fit(X,y)
```

```python
prd = mod.predict([[0,2]])
print(prd)
```
